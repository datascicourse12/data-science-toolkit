```python
#Part 1-Google Colab and Python
# 1. Create a new notebook and code the following
# 1. Create a numpy array of size 2 x 2

#numpy is a python library that allows you to work with numbers and large datasets etc
import numpy as np


array_2x2 = np.array ([
    [1,2],
    [3,4]
])

print("Array 2x2")
print(array_2x2)
```

```python
# 2. Create a random numpy array of size 3 x 4

#The first number is the row and second is the column
#So this is a grid of 3 rows and 4 columns

#this allows you to make an array of shape 3,4 and fill it with random numbers. Your numbers will be float
random3x4 = np.random.rand(3,4)

print("Random 3x4 array")
print(random3x4)
```

```python
# 3. Create a numpy array of ones and a numpy array of zeroes
ones_array = np.ones((3, 3))
zeros_array = np.zeros((3, 3))

print("Array of Ones:")
print(ones_array)

print("\nArray of Zeros:")
print(zeros_array)
```

```python
# 4. Create a random numpy array of size 6x3 and transpose it
#when you transpose something you switch the rows and columns
random_6x3 = np.random.rand(6, 3)

transpose_array = random_6x3.T

print("Original Array (6x3):")
print(random_6x3)

print("\nTransposed Array (3x6):")
print(transpose_array)
```

```python

# 5. Create two random numpy array of size n x p and p x m and then compute the dot product
#so one array with n rows and p columsn
#And then another array with p number of rows and m number of columns
#The dot product then

n = 3
p = 4
m = 2

array_A = np.random.rand(n, p)
array_B = np.random.rand(p, m)

dot_product = np.dot(array_A, array_B)
#3x4 and 4x2 so dot becomes 3*4=12 and 4*2=8 12x8=96
#your two inner numbers must match so in this case 4
#you are multiplying the first row in a with the first column in b add all together and you will get 1.2
print("Array A:")
print(array_A)

print("\nArray B:")
print(array_B)

print("\nDot Product:")
print(dot_product)

print("\nResult Shape:", dot_product.shape)



#second way of doing it
print("Option 2")
print(array_A @ array_B)
```
